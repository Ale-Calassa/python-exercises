import json
import sqlite3

def importar_json_para_sqlite(categoria):
    caminho = f"perguntas/{categoria.lower()}.json"
    with open(caminho, "r", encoding="utf-8") as f:
        perguntas = json.load(f)

    conn = sqlite3.connect("quiz.db")
    cursor = conn.cursor()

    for p in perguntas:
        cursor.execute("""
        INSERT INTO perguntas (categoria, nivel, pergunta, opcoes, resposta, extra)
        VALUES (?, ?, ?, ?, ?, ?)
        """, (
            categoria,
            p["nivel"],
            p["pergunta"],
            json.dumps(p["opcoes"]),  # salva lista como JSON
            p["resposta"],
            p["extra"]
        ))

    conn.commit()
    conn.close()

```python
import random
import sqlite3
import json

def carregar_perguntas_sqlite(categoria, nivel=None):
    conn = sqlite3.connect("/home/jovyan/work/QuizPython03/quiz.db")
    cursor = conn.cursor()

    if nivel:
        cursor.execute("""
        SELECT pergunta, opcoes, resposta, extra FROM perguntas
        WHERE LOWER(categoria) = LOWER(?) AND LOWER(nivel) = LOWER(?)
        """, (categoria, nivel))
    else:
        cursor.execute("""
        SELECT pergunta, opcoes, resposta, extra FROM perguntas
        WHERE LOWER(categoria) = LOWER(?)
        """, (categoria,))

    rows = cursor.fetchall()
    conn.close()

    perguntas = []
    for pergunta, opcoes_json, resposta, extra in rows:
        perguntas.append({
            "pergunta": pergunta,
            "opcoes": json.loads(opcoes_json),
            "resposta": resposta,
            "extra": extra
        })

    random.shuffle(perguntas)
    return perguntas
```

```python
# Função para rodar o quiz
import time
import csv
import os

def rodar_quiz(perguntas, jogador, tempo_limite=15):
    pontuacao = 0
    respostas_certas = []
    respostas_erradas = []

    for i, q in enumerate(perguntas):
        print(f"\nPergunta {i+1}: {q['pergunta']}")
        for idx, opcao in enumerate(q['opcoes']):
            print(f"{idx}. {opcao}")

        inicio = time.time()
        resposta = None

        while True:
            try:
                resposta = int(input(f"Digite o número da resposta (tempo limite: {tempo_limite}s): "))
                if resposta < 0 or resposta >= len(q['opcoes']):
                    print("⚠️ Opção inválida. Tente novamente.")
                    continue
                break
            except ValueError:
                print("⚠️ Entrada inválida. Digite um número.")
            if time.time() - inicio > tempo_limite:
                print("⏱️ Tempo esgotado!")
                resposta = None
                break

        if resposta == q['resposta']:
            print("✅ Correto!", q['extra'])
            pontuacao += 1
            respostas_certas.append(q['pergunta'])
        else:
            print(f"❌ Errado! A resposta correta era: {q['opcoes'][q['resposta']]}")
            respostas_erradas.append(q['pergunta'])

    print(f"\nPontuação final de {jogador}: {pontuacao}/{len(perguntas)}")
    salvar_resultado_csv(jogador, pontuacao, len(perguntas))
    return pontuacao
```

```python
# Ranking
def salvar_resultado_csv(jogador, pontuacao, total):
    arquivo = '/home/jovyan/work/QuizPython03/ranking.csv'
    existe = os.path.isfile(arquivo)
    with open(arquivo, 'a', newline='', encoding='utf-8') as f:
        writer = csv.writer(f)
        if not existe:
            writer.writerow(['Jogador', 'Pontuação', 'Total de Perguntas'])
        writer.writerow([jogador, pontuacao, total])

def mostrar_ranking():
    try:
        with open('/home/jovyan/work/QuizPython03/ranking.csv', 'r', encoding='utf-8') as f:
            reader = csv.reader(f)
            next(reader)  # pular cabeçalho
            ranking = sorted(reader, key=lambda x: int(x[1]), reverse=True)
            print("\n🏆 Ranking:")
            for i, linha in enumerate(ranking[:5]):
                print(f"{i+1}. {linha[0]} - {linha[1]}/{linha[2]}")
    except FileNotFoundError:
        print("Nenhum ranking disponível ainda.")
```

```python
# Execução do Quiz
jogador = input("Digite seu nome: ")
categoria = input("Escolha uma categoria (programacao, web): ")
nivel = input("Escolha o nível (Fácil, Médio, Difícil): ")

perguntas = carregar_perguntas_sqlite(categoria, nivel)

if not perguntas:
    print("⚠️ Nenhuma pergunta encontrada.")
else:
    pontuacao = rodar_quiz(perguntas, jogador)
    mostrar_ranking()
```
